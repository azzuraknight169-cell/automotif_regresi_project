Tujuan: memeriksa dataset sebelum membangun model regresi linear sederhana untuk hubungan kapasitas mesin dan efisiensi BBM jalan raya.

```python
import pandas as pd

df = pd.read_csv("mpg.csv")
print("Ukuran dataset:", df.shape)
print("\nLima baris pertama:")
display(df.head())
print("\nNama kolom:")
print(df.columns.tolist())
```

```python
data = df[["displ", "hwy"]].copy()
print("Jumlah nilai kosong per variabel:")
print(data.isna().sum())
print("\nStatistik ringkas:")
print(data.describe())
```

```python
import matplotlib.pyplot as plt

plt.scatter(data["displ"], data["hwy"], alpha=0.7)
plt.xlabel("Kapasitas mesin (liter)")
plt.ylabel("Efisiensi jalan raya (MPG)")
plt.title("Kapasitas Mesin vs Efisiensi BBM")
plt.grid(alpha=0.25)
plt.show()
```

**Sumber data:** dokumentasi dataset mpg dari plotnine/ggplot2: https://plotnine.org/reference/mpg.html. Dataset merupakan subset data konsumsi bahan bakar dari EPA AS untuk mobil tahun 1999 dan 2008.
